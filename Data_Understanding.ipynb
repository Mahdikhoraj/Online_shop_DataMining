# Number and pecentage of duplicated values

```python
duplicate_mask = df.duplicated()

duplicate_count = duplicate_mask.sum()

duplicate_percent = duplicate_count / len(df) * 100

print("Duplicate count:", duplicate_count)
print("Duplicate percent:", round(duplicate_percent, 2), "%")
```

# برسی فاکتور های لغو شده
<p>در این دیتا ست فاکتور های که با C شروع میشوند لغو شده هستند.<p>

چند ردیف مربوط به فاکتورهای لغوشده است؟

چند درصد از کل داده‌ها را تشکیل می‌دهند؟

آیا تمام ردیف‌های لغوشده دارای Quantity منفی هستند؟

```python
cancel_mask = df["InvoiceNo"].astype("string").str.startswith("C")

cancelled = df[cancel_mask]

print("Cancelled rows:", len(cancelled))

print("Cancelled percent:", round(
    len(cancelled) / len(df) * 100, 2
), "%")

print("Negative Quantity:", (cancelled["Quantity"] < 0).sum())

print("Non-negative Quantity:", (cancelled["Quantity"] >= 0).sum())
```

# برسی قیمت های صفر و منفی

```python
zero_price = df[df["UnitPrice"] == 0]

negative_price = df[df["UnitPrice"] < 0]

print("Zero prices:", len(zero_price))
print("Negative prices:", len(negative_price))

display(zero_price.head(10))
display(negative_price.head(10))
```

# برسی مقادیر گم شده در Costumer id

```python
missing_customer = df[df["CustomerID"].isna()]

print("Missing CustomerID:", len(missing_customer))

print(
    missing_customer["Country"]
    .value_counts()
)

display(missing_customer.head(20))
```

# برسی مقادیر گم شده در Description

```python
missing_description = df[df["Description"].isna()]

print("Missing Description:", len(missing_description))

print(
    "Missing CustomerID among them:",
    missing_description["CustomerID"].isna().sum()
)

print(
    "Zero UnitPrice among them:",
    (missing_description["UnitPrice"] == 0).sum()
)

print(
    "Negative UnitPrice among them:",
    (missing_description["UnitPrice"] < 0).sum()
)

display(missing_description.head(20))
```
