<div style='border-right:4px solid #4a7fb5;background:rgba(74,127,181,0.12);padding:10px 16px;margin:12px 0;border-radius:6px;font-size:15.5px;line-height:1.9'><b>نکته:</b> در متن سوالات از «بیمارستان» نام برده شده بود، اما سناریوی این تسک یک <b>فروشگاه آنلاین بین‌المللی</b> است. همه‌ی پاسخ‌ها بر اساس همین سناریو نوشته شده‌اند.</div>

<h1 style='font-size:25px;line-height:1.6;margin:18px 0 10px;padding-bottom:6px;border-bottom:2px solid rgba(74,127,181,0.7)'>بخش ۱: مسئله‌ی کسب‌وکار</h1>
<h2 style='font-size:20px;line-height:1.6;margin:20px 0 8px;color:#4a7fb5'>سوال ۱: مسئله‌ی اصلی کسب‌وکار را با زبان ساده توضیح دهید.</h2>
<h3 style='font-size:17px;line-height:1.6;margin:16px 0 6px;font-weight:700'>چرا این سوال را می‌پرسیم؟</h3>
<p style='margin:8px 0;font-size:16px;line-height:1.9'>چون اگر نتوانیم مسئله را <b>بدون اصطلاح فنی</b> و برای یک مدیر غیرفنی توضیح دهیم، یعنی هنوز آن را نفهمیده‌ایم. همه‌ی تصمیم‌های بعدی (داده، ویژگی، مدل) از همین جمله ساخته می‌شوند.</p>
<h3 style='font-size:17px;line-height:1.6;margin:16px 0 6px;font-weight:700'>پاسخ</h3>
<p style='margin:8px 0;font-size:16px;line-height:1.9'>یک فروشگاه آنلاین متوجه شده بعضی مشتریان قدیمی‌اش <b>کم‌کم خریدشان را کم می‌کنند</b> و بعضی دیگر <b>بی‌صدا و ناگهانی</b> دیگر برنمی‌گردند. مشکل اینجاست که شرکت <b>بعد از رفتن</b> مشتری متوجه می‌شود و دیگر دیر شده است.</p>
<p style='margin:8px 0;font-size:16px;line-height:1.9'>شرکت می‌خواهد <b>قبل از رفتن مشتری</b>، کسانی را که احتمال رفتنشان زیاد است پیدا کند و با یک اقدام مناسب (تخفیف، یادآوری، پیشنهاد محصول) سعی کند نگهشان دارد.</p>
<p style='margin:8px 0;font-size:16px;line-height:1.9'>یک محدودیت واقعی هم وجود دارد: بودجه‌ی تماس فقط برای <b>۵۰۰ مشتری</b> است. پس باید <b>درست انتخاب کنیم</b> که با چه کسانی تماس بگیریم.</p>
<h3 style='font-size:17px;line-height:1.6;margin:16px 0 6px;font-weight:700'>مثال ساده</h3>
<p style='margin:8px 0;font-size:16px;line-height:1.9'>مثل یک پزشک خانواده که به‌جای صبر کردن تا بیمار حالش بد شود، با نگاه به علائم اولیه (کم‌شدن رفت‌وآمد، دیرتر آمدن) پیش‌دستی می‌کند. اینجا «علائم اولیه» همان <b>الگوی خرید مشتری</b> است.</p>

<h3 style='font-size:17px;line-height:1.6;margin:16px 0 6px;font-weight:700'>چرا می‌پرسیم؟</h3>
<p style='margin:8px 0;font-size:16px;line-height:1.9'>تا <b>ارزش پروژه</b> را نشان دهیم. اگر نبودِ پروژه هزینه‌ای نداشته باشد، اجرای آن هم توجیهی ندارد. این سوال «چرا باید پول و وقت خرج کنیم؟» را جواب می‌دهد.</p>
<h3 style='font-size:17px;line-height:1.6;margin:16px 0 6px;font-weight:700'>پاسخ</h3>
<ul style='padding:0 24px 0 0;margin:8px 0;font-size:16px;line-height:1.9'><li style='margin:4px 0'><b>از دست رفتن درآمد:</b> هر مشتری که می‌رود، خریدهای آینده‌اش هم می‌رود.</li><li style='margin:4px 0'><b>هزینه‌ی بالای جذب مشتری جدید:</b> جذب مشتری تازه معمولاً گران‌تر از نگه‌داشتن مشتری فعلی است.</li><li style='margin:4px 0'><b>تأخیر در واکنش:</b> شرکت فقط بعد از ریزش متوجه می‌شود و فرصت اقدام ندارد.</li><li style='margin:4px 0'><b>هدررفت بودجه‌ی بازاریابی:</b> بدون مدل، تماس و تخفیف <b>تصادفی یا به همه</b> داده می‌شود، یعنی هم هزینه‌ی زیاد، هم اثر کم.</li><li style='margin:4px 0'><b>نبود دید رفتاری:</b> شرکت نمی‌داند کدام گروه‌ها در معرض خطرند و چرا.</li></ul>

<p style='margin:8px 0;font-size:16px;line-height:1.9'><b>Accuracy (دقت کلی)</b> = تعداد پیش‌بینی‌های درست ÷ کل پیش‌بینی‌ها</p>
<h3 style='font-size:17px;line-height:1.6;margin:16px 0 6px;font-weight:700'>پاسخ: <b>نه، کافی نیست.</b> (خود تسک هم صریحاً این را می‌گوید.)</h3>
<p style='margin:8px 0;font-size:16px;line-height:1.9'><b>دلیل با مثال:</b> فرض کنید از هر ۱۰۰ مشتری فقط ۱۰ نفر ریزش می‌کنند. مدلی که <b>همه را «غیر ریزشی»</b> پیش‌بینی کند:</p>
<ul style='padding:0 24px 0 0;margin:8px 0;font-size:16px;line-height:1.9'><li style='margin:4px 0'>Accuracy = ۹۰٪ (به‌ظاهر عالی!)</li><li style='margin:4px 0'>ولی <b>هیچ ریزشی را پیدا نکرده</b> (Recall = ۰٪) و از نظر کسب‌وکار بی‌فایده است.</li></ul>
<p style='margin:8px 0;font-size:16px;line-height:1.9'>پس Accuracy می‌تواند <b>گمراه‌کننده</b> باشد، به‌خصوص وقتی کلاس‌ها نامتوازن‌اند. به‌جای آن از Precision، Recall، F1 و PR-AUC استفاده می‌کنیم.</p>
<div style='border-right:4px solid #4a7fb5;background:rgba(74,127,181,0.12);padding:10px 16px;margin:12px 0;border-radius:6px;font-size:15.5px;line-height:1.9'><b>توجه:</b> در این پروژه میزان عدم توازن به تاریخ مرجع بستگی دارد و باید در مرحله‌ی Data Understanding اندازه‌گیری شود، نه حدس زده شود.</div>
</div>

<div dir='rtl' style='text-align:right;font-family:Vazirmatn, Tahoma, Segoe UI, Arial, sans-serif;font-size:16px;line-height:1.9;max-width:900px'>
<h2 style='font-size:20px;line-height:1.6;margin:20px 0 8px;color:#4a7fb5'>سوال ۱۸: در این پروژه Precision مهم‌تر است یا Recall؟ چرا؟</h2>
<h3 style='font-size:17px;line-height:1.6;margin:16px 0 6px;font-weight:700'>چرا می‌پرسیم؟</h3>
<p style='margin:8px 0;font-size:16px;line-height:1.9'>چون این دو معیار با هم <b>تعادل</b> دارند: بالا بردن یکی معمولاً دیگری را پایین می‌آورد. باید بر اساس <b>هزینه‌ی هر نوع خطا</b> تصمیم بگیریم.</p>
<h3 style='font-size:17px;line-height:1.6;margin:16px 0 6px;font-weight:700'>دو نوع خطا</h3>
<ul style='padding:0 24px 0 0;margin:8px 0;font-size:16px;line-height:1.9'><li style='margin:4px 0'><b>False Positive (هشدار اشتباه):</b> مشتری‌ای که نمی‌رفت را پرریسک اعلام کردیم ← یک تماس یا تخفیف بی‌دلیل هزینه شد.</li><li style='margin:4px 0'><b>False Negative (از دست دادن):</b> مشتری‌ای که می‌رفت را پیدا نکردیم ← مشتری را بدون تلاش از دست دادیم.</li></ul>
<h3 style='font-size:17px;line-height:1.6;margin:16px 0 6px;font-weight:700'>پاسخ (با دو لایه)</h3>
<p style='margin:8px 0;font-size:16px;line-height:1.9'><b>در حالت کلی churn، Recall اغلب مهم‌تر است</b>، چون از دست دادن یک مشتری ارزشمند معمولاً گران‌تر از یک تماس اضافی است.</p>
<p style='margin:8px 0;font-size:16px;line-height:1.9'><b>ولی در این پروژه یک قید خاص داریم: فقط ۵۰۰ تماس.</b> چون ظرفیت ثابت است، هر تماس برای یک مشتری اشتباه یعنی یک تماس از دست رفته. پس برای <b>کیفیت لیست ۵۰۰ نفره</b>، <b>Precision@500</b> معیار عملیاتی اصلی است.</p>
<table dir='rtl' style='width:100%;border-collapse:collapse;margin:12px 0;font-size:15px;line-height:1.8'><tr><th style='border:1px solid rgba(128,128,128,0.45);padding:8px 12px;text-align:right;vertical-align:top;background:rgba(128,128,128,0.2);font-weight:700'>سطح</th><th style='border:1px solid rgba(128,128,128,0.45);padding:8px 12px;text-align:right;vertical-align:top;background:rgba(128,128,128,0.2);font-weight:700'>معیار مهم‌تر</th><th style='border:1px solid rgba(128,128,128,0.45);padding:8px 12px;text-align:right;vertical-align:top;background:rgba(128,128,128,0.2);font-weight:700'>دلیل</th></tr><tr><td style='border:1px solid rgba(128,128,128,0.45);padding:8px 12px;text-align:right;vertical-align:top'>ارزیابی کلی مدل</td><td style='border:1px solid rgba(128,128,128,0.45);padding:8px 12px;text-align:right;vertical-align:top'>Recall و PR-AUC</td><td style='border:1px solid rgba(128,128,128,0.45);padding:8px 12px;text-align:right;vertical-align:top'>تا مشتریان در حال رفتن را از دست ندهیم</td></tr><tr><td style='border:1px solid rgba(128,128,128,0.45);padding:8px 12px;text-align:right;vertical-align:top'>انتخاب ۵۰۰ نفر</td><td style='border:1px solid rgba(128,128,128,0.45);padding:8px 12px;text-align:right;vertical-align:top'><b>Precision@500</b></td><td style='border:1px solid rgba(128,128,128,0.45);padding:8px 12px;text-align:right;vertical-align:top'>چون بودجه ثابت است و هر تماس باید به مشتری پرریسک برسد</td></tr><tr><td style='border:1px solid rgba(128,128,128,0.45);padding:8px 12px;text-align:right;vertical-align:top'>مقایسه‌ی متعادل</td><td style='border:1px solid rgba(128,128,128,0.45);padding:8px 12px;text-align:right;vertical-align:top'>F1</td><td style='border:1px solid rgba(128,128,128,0.45);padding:8px 12px;text-align:right;vertical-align:top'>وقتی هر دو مهم‌اند</td></tr></table>
<div style='border-right:4px solid #4a7fb5;background:rgba(74,127,181,0.12);padding:10px 16px;margin:12px 0;border-radius:6px;font-size:15.5px;line-height:1.9'><b>جمع‌بندی:</b> هیچ‌کدام به‌تنهایی «درست مطلق» نیست. انتخاب نهایی به <b>هزینه‌ی تماس</b> و <b>ارزش مشتری</b> بستگی دارد، که دقیقاً در تحلیل حساسیت مرحله‌ی کمپین بررسی می‌شود.</div>

<h2 style='font-size:20px;line-height:1.6;margin:20px 0 8px;color:#4a7fb5'>سوال ۱۹: موفقیت مدل را چگونه تعریف می‌کنید؟</h2>
<h3 style='font-size:17px;line-height:1.6;margin:16px 0 6px;font-weight:700'>چرا می‌پرسیم؟</h3>
<p style='margin:8px 0;font-size:16px;line-height:1.9'>موفقیت <b>مدل</b> با موفقیت <b>کسب‌وکار</b> فرق دارد. اینجا فقط می‌گوییم مدل چه شرایطی را باید داشته باشد تا «قابل استفاده» باشد.</p>
<h3 style='font-size:17px;line-height:1.6;margin:16px 0 6px;font-weight:700'>پاسخ</h3>
<p style='margin:8px 0;font-size:16px;line-height:1.9'>مدل موفق است اگر <b>همه‌ی</b> شرایط زیر برقرار باشد:</p>
<ol style='padding:0 24px 0 0;margin:8px 0;font-size:16px;line-height:1.9'><li style='margin:4px 0'><b>بهتر از خط مبنا:</b> از مدل ساده‌ی RFM در PR-AUC و Precision@500 بهتر باشد. (اگر نباشد، پیچیدگی اضافی ارزشی ندارد.)</li><li style='margin:4px 0'><b>بهتر از تصادفی:</b> در ۵۰۰ نفر اول، ریزشی‌های بیشتری از انتخاب تصادفی پیدا کند.</li><li style='margin:4px 0'><b>ارزیابی درست:</b> نتیجه روی <b>داده‌ی زمانی آینده</b> (نه تقسیم تصادفی) به‌دست آمده باشد.</li><li style='margin:4px 0'><b>بدون نشت:</b> هیچ ویژگی‌ای اطلاعات آینده را آشکار نکند.</li><li style='margin:4px 0'><b>پایدار:</b> عملکرد در تاریخ‌های مرجع مختلف تقریباً ثابت باشد.</li><li style='margin:4px 0'><b>قابل تفسیر:</b> عوامل مؤثر قابل توضیح برای مدیریت باشند.</li><li style='margin:4px 0'><b>سودآور در تحلیل اقتصادی:</b> با فرض‌های شفاف، سود مورد انتظار کمپین مثبت باشد.</li></ol>
<div style='border-right:4px solid #4a7fb5;background:rgba(74,127,181,0.12);padding:10px 16px;margin:12px 0;border-radius:6px;font-size:15.5px;line-height:1.9'><b>یادآوری:</b> مطابق متن تسک، <b>هدف پروژه رسیدن به بیشترین دقت نیست.</b> مدلی با عدد کمی پایین‌تر ولی ارزیابی درست و تصمیم قابل دفاع، بهتر از مدل با عدد بالا و نشت اطلاعات است.</div>
